```python
input_rdd = sc.parallelize({1,2,3,4,5,1,2,1})

output_rdd = input_rdd.map(lambda x: x*10)
print(output_rdd.collect())
```

```python
# generate an output rdd by collecting if ele >= 3
output_rdd = input_rdd.filter( lambda x: x >=3 )
print(output_rdd.collect())
```

```python
list =['pyspark session', 'is started', 'rdd topic']
rdd = sc.parallelize(list)
map_rdd= rdd.map(lambda str : str.split(' '))
print(map_rdd.collect())

flatmap_rdd= rdd.flatMap(lambda str : str.split(' '))
print(flatmap_rdd.collect())
```

```python
statement  =['pyspark is started session is started rdd topic']
# count of word
rdd= sc.parallelize(statement)
words_rdd = rdd.flatMap( lambda x: x.split(' '))
print(words_rdd.collect())

rdd1= words_rdd.map(lambda word: (word, 1))
print(rdd1.collect())

count_rdd = rdd1.reduceByKey(lambda x, y : x+y)
print(count_rdd.collect())

result = words_rdd.countByValue()
print(result)
print(type(result))
```

```python
rdd= sc.textFile(r"C:\Users\hp\PycharmProjects\PythonProject1\Data\cust.csv")
print(rdd.collect())

columns = rdd.first()

# new rdd whose id >2
lines = rdd.filter(lambda x: x!= columns) \
            .map(lambda x: x.split(','))
print(lines.collect())

result = lines.filter(lambda row: int(row[0]) >2)
print(result.collect())
```
