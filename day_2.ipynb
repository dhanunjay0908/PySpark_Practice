```python
xml_df = spark.read.format("xml").option("rowTag", "book").load("book.xml")
# xml_df.show()
xml_df.printSchema()
# xml_df.select("author").show(20, False)

# how many no of books each author published
# select author, count(title) as book_count from books group by author having book_count >2

xml_df.groupBy("author").agg(count("title").alias("book_count")).filter(col("book_count") > 2).show()
```
